Cell 1: Library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

Cell 2: Data Sintetis

```python
np.random.seed(2008)
n_samples = 100

x1 = np.random.rand(n_samples) * 2
x2 = np.random.rand(n_samples) * 3

noise = np.random.rand(n_samples) * 1.5

#pengaruh linear: Y = 6 + 2.5*x1 - 2 *x2 + noise
Y = 6 + 2.5 * x1 - 2 * x2 + noise

#Menggabungkan menjadi data frame
df = pd.DataFrame({'x1' : x1, 'x2' : x2, 'Y' : Y})
display(df.head())
```

Cell 3: Model Fit

```python
X = df[['x1', 'x2']]
y = df[['Y']]

#Membagi trainingn 80% testing 20%
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2008)

#Melakukan fitting model
model = LinearRegression()
model.fit(X_train, y_train)

#Melihat intercept dan koefisien
print(f"Intercept (konstanta): {model.intercept_[0]:.4f}")
print(f"Koefisien x1: {model.coef_[0][0]:.4f}")
print(f"Koefisien x2: {model.coef_[0][1]:.4f}")
```

Cell 4: Prediksi dan Matriks

```python
#Prediksi
y_pred = model.predict(X_test)

#Matriks
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

#Print Matriks
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R-squared (R2): {r2:.4f}")
```

Cell 5: Visualisasi

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(x=y_test.squeeze(), y=y_pred.flatten(), color = 'blue', alpha= 0.7)
plt.plot([y_test.min().item(), y_test.max().item()], [y_test.min().item(), y_test.max().item()], color= 'red', linestyle='--', lw=2)
plt.title('Nilai Aktual vs Prediksi')
plt.xlabel('Nilai Aktual (Y)')
plt.ylabel('Prediksi (Y_topi)')
plt.grid(True)
plt.show()
```
